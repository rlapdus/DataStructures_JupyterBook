## Inorder Traversal

```python
def InOrder(node):
  if node is None: return None
  InOrder(node.left) # explore left subtree
  print(node.data, end=' ') # print element in the node
  InOrder(node.right) # explore right subtree

print("Inorder traversal of the binary tree is:")
InOrder(tree)
```

## Pre/Postorder Traversal

```python
# Preorder: node → left → right
def PreOrder(node):
    if node is None:
        return None

    print(node.data, end=' ')
    PreOrder(node.left)
    PreOrder(node.right)




# Postorder: left → right → node
def PostOrder(node):
    if node is None:
        return None

    PostOrder(node.left)
    PostOrder(node.right)
    print(node.data, end=' ')


# Create tree
tree = Node(1)
tree.left = Node(2)
tree.right = Node(3)
tree.left.left = Node(4)
tree.left.left.left = Node(5)
tree.right.left = Node(6)
tree.right.right = Node(7)
tree.right.right.left = Node(8)
tree.right.right.right = Node(9)

# Example
print("Preorder traversal of the binary tree is:")
#
PreOrder(tree)
print()

print("Postorder traversal of the binary tree is:")
#
PostOrder(tree)
print()
```

### Add a print method in BinarySearchTree module

```python
class BinarySearchTree:
  def __init__(self):
    self.root = None

	# Search
  def search(self, x) -> TreeNode:
    return self.searchItem(self.root, x)

  def searchItem(self, tNode:TreeNode, x) -> TreeNode:
    if (tNode == None):
      return None
    elif (x == tNode.data):
      return tNode
    elif (x < tNode.data):
      return self.searchItem(tNode.left, x)
    else:
      return self.searchItem(tNode.right, x)

	# Insertion
  def insert(self, newItem):
    self.root = self.insertItem(self.root, newItem)

  def insertItem(self, tNode:TreeNode, newItem) -> TreeNode:
    if (tNode == None):
      tNode = TreeNode(newItem, None, None)
    elif (newItem < tNode.data):  # branch left
      tNode.left = self.insertItem(tNode.left, newItem)
    else: # branch right
      tNode.right = self.insertItem(tNode.right, newItem)
    return tNode

	# Deletion
  def delete(self, x):
    self.root = self.deleteItem(self.root, x)

  def deleteItem(self, tNode:TreeNode, x) -> TreeNode:
    if (tNode == None): # return none if item not found
      return None
    elif (x == tNode.data):	# if x is found, delete that node
      tNode = self.deleteNode(tNode)
    elif (x < tNode.data): # if x smaller, move to left child
      tNode.left = self.deleteItem(tNode.left, x)
    else: # if x bigger, move to right child
      tNode.right = self.deleteItem(tNode.right, x)
    return tNode # tNode: resulting node

  def deleteNode(self, tNode:TreeNode) -> TreeNode:
		#     1. tNode is leaf node
		#     2. tNode has only one children
		#     3. tNode has two children
    if tNode.left == None and tNode.right == None: # case 1 (leaf node)
      return None
    elif tNode.left == None:  # case 2 (only right child)
      return tNode.right
    elif tNode.right == None: # case 2 (onlt left child)
      return tNode.left
    else: # case 3 (two children)
      (rtnItem, rtnNode) = self.deleteMinItem(tNode.right)
      tNode.data = rtnItem
      tNode.right = rtnNode
      return tNode  # tNode survived

  def deleteMinItem(self, tNode:TreeNode) -> tuple:
    if tNode.left == None:
      # found min at tNode
      return (tNode.data, tNode.right)
    else: # branch left
      (rtnItem, rtnNode) = self.deleteMinItem(tNode.left)
      tNode.left = rtnNode
      return (rtnItem, tNode)

  def isEmpty(self) -> bool:
    return self.root == self.NIL

  def clear(self):
    self.root = self.NIL

  def printBST(self):
    InOrder(self.root)
    print()


tree = BinarySearchTree()
tree.insert(8); tree.insert(3); tree.insert(10);
tree.insert(1); tree.insert(6); tree.insert(14);
tree.insert(4); tree.insert(7); tree.insert(13)

tree.printBST()
```
